```python
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
import random
```

```python
df = pd.read_csv("Churn.txt")

def transform(x):
    if x == 'True.':
        return 1
    else:
        return 0

def transform2(x):
    if x =='yes':
        return 1
    else:
        return 0

df["Int'l Plan"] = df["Int'l Plan"].apply(transform2)
df['VMail Plan'] = df['VMail Plan'].apply(transform2)

sub_df = df[['Day Charge','Eve Charge','Night Charge','CustServ Calls', 'Churn?', "Int'l Plan", "VMail Plan"]].copy()

sub_df['target'] = sub_df['Churn?'].apply(transform)

# Je sais pas pk mais le prof a oublié les dimensions demandées dans l'exercice d'avant
#X = sub_df[[ 'Day Charge','Eve Charge','Night Charge','CustServ Calls', "Int'l Plan", "VMail Plan"]]

X = sub_df[[ 'Day Charge','Eve Charge','Night Charge','CustServ Calls']]
y = sub_df['target']

model = LogisticRegression()
model.fit(X, y)
```

```python
other_df = pd.DataFrame()
other_df['target'] = y
other_df['proba'] = model.predict_proba(X)[:,1]

roc_auc_score(other_df['target'], other_df['proba'])
```

```python
max_auc = 0
max_features = []
best_auc = 0
best_features = []

for feature in X.columns:
    selected_X = X[[feature]]

    model = LogisticRegression()
    model.fit(selected_X, y)

    other_df = pd.DataFrame()
    other_df['target'] = y
    other_df['proba'] = model.predict_proba(selected_X)[:,1]

    auc = roc_auc_score(other_df['target'], other_df['proba'])

    if auc > best_auc:
        best_auc = auc
        best_features = [feature]

    print(f"Feature: {feature}, AUC: {auc}")

max_auc = best_auc
max_features = best_features
print(f"\nBest & max features: {max_features}, Best AUC: {max_auc}")
```

```python
best_auc = 0
best_features = []

for feature1 in X.columns:
    for feature2 in X.columns:
        if feature1 != feature2:
            selected_X = X[[feature1, feature2]]

            model = LogisticRegression()
            model.fit(selected_X, y)

            other_df = pd.DataFrame()
            other_df['target'] = y
            other_df['proba'] = model.predict_proba(selected_X)[:,1]

            auc = roc_auc_score(other_df['target'], other_df['proba'])

            if auc > best_auc:
                best_auc = auc
                best_features = [feature1, feature2]

            print(f"Features: {feature1}, {feature2}, AUC: {auc}")

if best_auc > max_auc:
    max_auc = best_auc
    max_features = best_features
print(f"\nBest features: {best_features}, Best AUC: {best_auc}")
print(f"\nMax features is now : {max_features}, Best AUC: {max_auc}")
```
