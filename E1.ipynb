Montrer que les formules présentées dans le slide précédent correspondent bien à une courbe en S, confinée en ordonnée à l’intervalle [0,1].

```python
from math import *
import matplotlib.pyplot as plt
import pandas as pd
```

Définition de la sigmoïde

```python
# Valeurs arbitraires
b0 = 1
b1 = 10

# Les formules du slide 19
def pi(x):
    return (1) / (1 + exp(-g(x)))

def g(x):
    return b0 + b1 * x
```

On crée la courbe sur [-10;10]

```python
data_df = pd.DataFrame()

data_df['x'] = range(-10,10+1)

# Créer les y en fonction des x
data_df['y'] = data_df['x'].apply(lambda i:pi(i))

data_df
```

On affiche visuellement

```python
plt.plot(data_df['x'], data_df['y'], 'o-')
plt.hlines(y=0.5,xmin=-10,xmax=+10,alpha=0.5)
plt.vlines(x=0.0,ymin=0,ymax=1,alpha=0.5)
plt.title(r'Logistic curve with $\beta_0={}$ $\beta_1={}$'.format(b0,b1))
plt.xlabel('x')
plt.ylabel(r'$\pi(x)$')
plt.show()
```
